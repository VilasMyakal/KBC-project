```python
def matmultiplicate(mat1,mat2):

  p = len(mat1)
  r = len(mat2)
  q1 = len(mat1[0])
  q2 = len(mat2[0])

  if q1 != q1 :
    print("DImension Mismatched!!")
    return

  mat3 = []

  for i in range(p):
    temp = []
    for j in range(r):
      temp.append(0)
    mat3.append(temp)

  for i in range(p):
    for j in range(r):
      for k in range(q1):
        mat3[i][j] += mat1[i][k] * mat2[k][j]


  return mat3
```

```python
mat1 = [[10,20],[10,20]]
mat2 = [[10,20],[10,20]]

print("Maltiplication: ",matmultiplicate(mat1,mat2))
```
